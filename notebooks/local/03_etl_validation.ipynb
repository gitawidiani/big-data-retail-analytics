## 1. Validasi data Parquet
Pemeriksaan ini menjadi bukti bahwa hasil ETL dapat dibaca ulang, tidak kehilangan ID penting, dan tetap mempunyai nilai bisnis yang masuk akal.

```python
fact = spark.read.parquet(str(paths.parquet_dir))
validation = fact.agg(
    F.count('*').alias('rows'),
    F.countDistinct('receipt_id').alias('receipts'),
    F.sum(F.when(F.col('store_name').isNull(), 1).otherwise(0)).alias('missing_store'),
    F.sum(F.when(F.col('sku_name').isNull(), 1).otherwise(0)).alias('missing_sku'),
    F.sum(F.when(F.col('loyalty_segment').isNull(), 1).otherwise(0)).alias('missing_customer'),
    F.round(F.sum('total_value'), 2).alias('sales_value'),
    F.round(F.sum('gross_profit'), 2).alias('gross_profit'),
).first().asDict()
assert validation['missing_store'] == 0
assert validation['missing_sku'] == 0
assert validation['missing_customer'] == 0
validation
```

## 2. Membuat grafik laporan
Hanya hasil agregasi kecil yang dikonversi ke Pandas. Data transaksi penuh tetap diproses oleh Spark.

```python
monthly = (fact.groupBy('month').agg(F.sum('total_value').alias('sales_value'))
           .orderBy('month').toPandas())
stores = (fact.groupBy('store_name').agg(F.sum('total_value').alias('sales_value'))
          .orderBy(F.desc('sales_value')).limit(10).toPandas())
categories = (fact.groupBy('category').agg(F.sum('total_value').alias('sales_value'))
              .orderBy(F.desc('sales_value')).toPandas())
channels = (fact.groupBy('channel').agg(F.sum('total_value').alias('sales_value'))
            .orderBy(F.desc('sales_value')).toPandas())

monthly['month'] = pd.to_datetime(monthly['month'])
fig, ax = plt.subplots(figsize=(12, 5))
ax.plot(monthly['month'], monthly['sales_value'], color='#2563eb', linewidth=2)
ax.set(title='Tren Nilai Penjualan Bulanan', xlabel='Bulan', ylabel='Nilai penjualan')
ax.grid(alpha=.25)
fig.tight_layout()
fig.savefig(paths.charts_dir / 'monthly_sales.png', dpi=160)
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 6))
axes[0].barh(stores['store_name'][::-1], stores['sales_value'][::-1], color='#0f766e')
axes[0].set_title('10 Toko dengan Penjualan Tertinggi')
axes[0].set_xlabel('Nilai penjualan')
axes[1].bar(categories['category'], categories['sales_value'], color='#ea580c')
axes[1].set_title('Penjualan per Kategori')
axes[1].set_ylabel('Nilai penjualan')
axes[1].tick_params(axis='x', rotation=35)
fig.tight_layout()
fig.savefig(paths.charts_dir / 'store_and_category_sales.png', dpi=160)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 5))
bars = ax.bar(channels['channel'], channels['sales_value'], color=['#2563eb', '#0f766e', '#ea580c'])
ax.set_title('Nilai Penjualan per Kanal')
ax.set_xlabel('Kanal')
ax.set_ylabel('Nilai penjualan')
ax.grid(axis='y', alpha=.2)
ax.bar_label(bars, labels=[f'{value / 1_000_000_000:.2f} miliar' for value in channels['sales_value']], padding=3)
fig.tight_layout()
fig.savefig(paths.charts_dir / 'channel_sales.png', dpi=160, bbox_inches='tight')
plt.show()
```

## 3. Perbandingan transaksi dengan dan tanpa promosi
Perbandingan dilakukan pada tingkat `receipt_id`, bukan baris produk. Satu struk dianggap memakai promosi jika minimal satu produknya memiliki `promo_id`. Hasil menunjukkan hubungan atau asosiasi, bukan bukti bahwa promosi menyebabkan perubahan.

```python
promotion = (
    spark.read.option('header', True).option('inferSchema', True)
    .csv(str(paths.validation_dir / 'promotion_comparison'))
    .orderBy('receipt_has_promo')
    .toPandas()
)
display(promotion)

promo = promotion.loc[promotion['receipt_has_promo'] == 1].iloc[0]
nonpromo = promotion.loc[promotion['receipt_has_promo'] == 0].iloc[0]
comparison_summary = pd.DataFrame({
    'metric': ['Unit per struk', 'Nilai per struk', 'Laba per struk', 'Margin laba'],
    'tanpa_promosi': [nonpromo['avg_units_per_receipt'], nonpromo['avg_value_per_receipt'],
                       nonpromo['avg_profit_per_receipt'], nonpromo['profit_margin_pct']],
    'dengan_promosi': [promo['avg_units_per_receipt'], promo['avg_value_per_receipt'],
                        promo['avg_profit_per_receipt'], promo['profit_margin_pct']],
})
comparison_summary['difference_pct'] = (
    comparison_summary['dengan_promosi'] / comparison_summary['tanpa_promosi'] - 1
) * 100
display(comparison_summary.round(2))
```

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 5))
labels = promotion['promotion_group']
colors = ['#64748b', '#2563eb']
metrics = [
    ('avg_units_per_receipt', 'Rata-rata Unit per Struk', 'Unit'),
    ('avg_value_per_receipt', 'Rata-rata Nilai per Struk', 'Nilai penjualan'),
    ('profit_margin_pct', 'Margin Laba', 'Persen'),
]
for ax, (column, title, ylabel) in zip(axes, metrics):
    bars = ax.bar(labels, promotion[column], color=colors)
    ax.set_title(title)
    ax.set_ylabel(ylabel)
    ax.grid(axis='y', alpha=.2)
    ax.bar_label(bars, fmt='%.2f', padding=3)
fig.tight_layout()
fig.savefig(paths.charts_dir / 'promotion_comparison.png', dpi=160, bbox_inches='tight')
plt.show()
```

## 4. Analisis risiko inventori
File flag dari pembuat dataset digunakan sebagai label referensi. Analisis ini menghitung banyaknya pasangan SKU–toko per jenis risiko.

```python
inventory = spark.read.option('header', True).option('inferSchema', True).csv(
    str(paths.source_dir / 'inventory_snapshot.csv')
)
flags = spark.read.option('header', True).option('inferSchema', True).csv(
    str(paths.source_dir / 'sku_inventory_flags.csv')
)
print(f'Baris snapshot inventori: {inventory.count():,}')
flag_counts = flags.groupBy('flag').count().orderBy(F.desc('count')).toPandas()
inventory_status = (
    inventory.withColumn(
        'stock_status',
        F.when(F.col('stock_on_hand') <= F.col('safety_stock'), 'Di bawah safety stock')
         .when(F.col('stock_on_hand') <= F.col('reorder_point'), 'Perlu reorder')
         .otherwise('Stok memadai'),
    )
    .groupBy('stock_status').count()
    .orderBy(F.desc('count')).toPandas()
)
print(f'Baris snapshot inventori: {inventory.count():,}')
display(flag_counts)
display(inventory_status)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))
flag_bars = axes[0].bar(flag_counts['flag'], flag_counts['count'], color=['#dc2626', '#f59e0b'])
axes[0].set_title('Label Risiko Inventori')
axes[0].set_ylabel('Jumlah SKU')
axes[0].bar_label(flag_bars, padding=3)
axes[0].grid(axis='y', alpha=.2)
status_colors = {
    'Di bawah safety stock': '#dc2626',
    'Perlu reorder': '#f59e0b',
    'Stok memadai': '#16a34a',
}
status_bars = axes[1].bar(
    inventory_status['stock_status'], inventory_status['count'],
    color=[status_colors[value] for value in inventory_status['stock_status']],
)
axes[1].set_title('Kondisi Stok per Toko dan SKU')
axes[1].set_ylabel('Jumlah pasangan toko–SKU')
axes[1].tick_params(axis='x', rotation=15)
axes[1].bar_label(status_bars, padding=3)
axes[1].grid(axis='y', alpha=.2)
fig.tight_layout()
fig.savefig(paths.charts_dir / 'inventory_risk.png', dpi=160, bbox_inches='tight')
plt.show()
```

## 5. Load PostgreSQL — opsional
Salin `.env.example` menjadi `.env`, isi koneksi PostgreSQL, lalu ubah `ENABLE_POSTGRES_LOAD=1`. Yang dimuat secara default hanya tabel agregasi sehingga proses ringan. Jangan menaruh kata sandi di notebook atau Git.

```python
if os.getenv('ENABLE_POSTGRES_LOAD', '0') == '1':
    from sqlalchemy import URL, create_engine, text

    url = URL.create(
        'postgresql+psycopg2',
        username=os.environ['POSTGRES_USER'],
        password=os.environ['POSTGRES_PASSWORD'],
        host=os.getenv('POSTGRES_HOST', 'localhost'),
        port=int(os.getenv('POSTGRES_PORT', '5432')),
        database=os.environ['POSTGRES_DB'],
    )
    engine = create_engine(url)
    with engine.begin() as conn:
        conn.execute(text('CREATE SCHEMA IF NOT EXISTS retail_analytics'))
    for folder in paths.validation_dir.iterdir():
        if folder.is_dir():
            table = spark.read.option('header', True).option('inferSchema', True).csv(str(folder))
            table.toPandas().to_sql(folder.name, engine, schema='retail_analytics',
                                  if_exists='replace', index=False)
    print('Tabel agregasi berhasil dimuat ke PostgreSQL.')
else:
    print('Load PostgreSQL dilewati. Atur ENABLE_POSTGRES_LOAD=1 jika database sudah siap.')
```
